```python
# create yaml file for the model
import torch
print("GPU available:", torch.cuda.is_available(), torch.cuda.get_device_name(0) if torch.cuda.is_available() else "")

from pathlib import Path
dataset_root = Path("/kaggle/input/padel-ball-yolo")  # confirm this matches your dataset's actual slug

yaml_content = (
    f"path: {dataset_root}\n"
    "train: images/train\n"
    "val: images/val\n"
    "names:\n"
    "  0: ball\n"
)
Path("/kaggle/working/ball_dataset.yaml").write_text(yaml_content)
print(open("/kaggle/working/ball_dataset.yaml").read())
```

```python
# script to train the model
from ultralytics import YOLO

model = YOLO("yolo26s.pt")

results = model.train(
    data="/kaggle/working/ball_dataset.yaml",
    imgsz=1280,
    epochs=22,
    time=10,
    patience=15,
    batch=-1,
    close_mosaic=15,
    save_period=5,
    project="/kaggle/working/runs",
    name="ball_yolo26s_v1",
    resume=False,
)
```
